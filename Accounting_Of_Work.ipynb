```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# Selecting destination file path

```python
data=pd.read_csv(r'/home/biswarup/Desktop/Accounting_of_Work - Sheet1.csv')
data.head(10)
#df=pd.DataFrame(data, columns=['x','y'])
```

# Fetching the datas

```python
df=pd.DataFrame(data, columns=['Hours'])
print(df.head(10))
```

# Making row X

```python
x=[]
i=0
while i!=len(df):
    x.append(i)
    i+=1
print(len(x))
```

# Making Visual Data

```python
plt.title('Day(s) vs Hour(s)')
plt.xlabel('Day(s)')
plt.ylabel('Hour(s)')
plt.plot(x,df)
plt.savefig("/home/biswarup/Downloads/output.png")
plt.show()
```
