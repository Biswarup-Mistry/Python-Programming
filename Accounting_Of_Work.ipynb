# Fetching the datas

```python
df=pd.DataFrame(data, columns=['Hours'])
print(df.head(10))
```

# Marking the X & Y

```python
y=[]
x=[]
i=0
for i in range(len(df)):
    y.append(df['Hours'][i])
print(len(y))
i=0
for i in range(len(y)):
    x.append(i)
print(len(x))
```

# Making Visual Data

```python
plt.title('Day(s) vs Hour(s)')
plt.xlabel('Day(s)')
plt.ylabel('Hour(s)')
plt.plot(x,y)
#plt.savefig("/home/biswarup/Downloads/output.png") #for linux
plt.savefig(r"C:\Users\Biswarup\Downloads\output.png")  # for windows
#C:\Users\Biswarup\Downloads\output.png
plt.show()
```
